## 1.1  Load the Dataset

```python
# Load raw CSV — path is relative to the notebooks/ directory
DATA_PATH = os.path.join('..', 'customer_churn.csv')

df = load_dataset(DATA_PATH)
print(f'Dataset loaded: {df.shape[0]} rows × {df.shape[1]} columns')
```

## 1.3  Dataset Shape & Columns

```python
print(f'Shape : {df.shape}')
print(f'\nColumns ({len(df.columns)}):')
for i, col in enumerate(df.columns, 1):
    print(f'  {i:>2}.  {col}')
```

## 1.4  Data Types

```python
print('Data Types:')
print(df.dtypes)
print(f'\nNumeric columns  : {list(df.select_dtypes(include="number").columns)}')
print(f'Categorical columns: {list(df.select_dtypes(include="object").columns)}')
```

## 1.5  Missing Values & Duplicates

```python
# Missing value analysis
missing_summary = pd.DataFrame({
    'Missing Count': df.isnull().sum(),
    'Missing %'    : (df.isnull().mean() * 100).round(2)
})
missing_summary = missing_summary[missing_summary['Missing Count'] > 0]

if missing_summary.empty:
    print('✅  No explicit NaN values detected in the dataset.')
    print('    Note: TotalCharges may contain whitespace strings — handled in cleaning.')
else:
    print(missing_summary)

print(f'\nDuplicate rows: {df.duplicated().sum()}')
```

## 1.6  Descriptive Statistics

```python
# Numeric columns summary
print('Numeric Feature Statistics:')
df.describe()
```

```python
# Categorical columns — unique value counts
print('Categorical Feature Unique Values:')
cat_cols = df.select_dtypes(include='object').columns
for col in cat_cols:
    print(f'  {col:<25}: {df[col].nunique()} unique — {df[col].unique().tolist()}')
```

## 1.7  Full Dataset Summary (via preprocessing module)

## 1.8  Target Variable Distribution

```python
churn_counts = df['Churn'].value_counts()
churn_percentages = df['Churn'].value_counts(normalize=True) * 100

print('Churn Distribution:')
summary = pd.DataFrame({'Count': churn_counts, 'Percentage (%)': churn_percentages.round(2)})
print(summary)
print('\n⚠  Class imbalance present — will be handled during model training (class_weight=balanced).')
```
